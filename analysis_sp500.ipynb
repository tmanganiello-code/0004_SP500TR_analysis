SP500 analysis with weekly adjusted data (SP500TR)

```python
import pandas as pd
#data souces
sp500_raw = pd.read_csv('sp500_weekly_adjusted.csv')
print(sp500_raw.head(2))
```

```python
#data cleaning
sp500 = sp500_raw.ffill()

#convert date in datetime format
sp500['Date'] = pd.to_datetime(sp500['Date'])
sp500_cle = sp500[['Adj Close','Date']]

#sort the value
sp500_cle = sp500_cle.sort_values(by='Date', ascending=True)
print(sp500_cle.iloc[[0, -1]])
```

Sp500 weekly retun analysis

```python
#sp500 weekly return
sp500_cle['Return'] = sp500_cle['Adj Close'].pct_change()

#preparing data
sp500_ret = sp500_cle[['Date', 'Return', 'Adj Close' ]]
sp500_ret = sp500_ret.dropna(subset=['Return'])
print(sp500_ret.iloc[[0, -1]])
```

```python
mean_sp500 = sp500_ret['Return'].mean()
median_sp500 = sp500_ret['Return'].median()
mode_sp500 = sp500_ret['Return'].mode().iloc[0] 
stdev_sp500 = sp500_ret['Return'].std()         
sterror_sp500 = stdev_sp500 / (len(sp500_ret['Return']) ** 0.5)

z_score = 1.96
ci_lower = mean_sp500 - (z_score * sterror_sp500)
ci_upper = mean_sp500 + (z_score * sterror_sp500)

print(
    f"For the S&P 500 weekly returns (including dividends and splits),\n" 
    f"the mean return is {mean_sp500 * 100:.2f}%(with a median of {median_sp500 * 100:.2f}%).\n"
    f"The data shows a sample standard deviation of {stdev_sp500 * 100:.2f}%, \n"
    f"resulting in a standard error of the mean of {sterror_sp500 * 100:.4f}%. \n"
    f"We are 95% confident that the true population mean lies between \n"
    f"{ci_lower * 100:.2f}% and {ci_upper * 100:.2f}% (95% Confidence Interval).\n"
)
```

Annual yields from 1993 to 2025

```python
sp500_cle['Year'] = sp500_cle['Date'].dt.year

annual_yield = (sp500_cle.groupby('Year')['Return'].apply(lambda x: (1 + x).prod() - 1)*100)
print(annual_yield)
```
