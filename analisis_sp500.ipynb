SP500 analysis with weekly adjusted data (SP500TR)

Data Collection

```python
import pandas as pd
#data souces
sp500_raw = pd.read_csv('sp500_weekly_adjusted.csv')
```

Data Cleaning

```python
sp500 = sp500_raw.ffill()

#convert date in datetime format
sp500['Date'] = pd.to_datetime(sp500['Date'])
sp500_cle = sp500[['Adj Close','Date']]

#sort the value
sp500_cle = sp500_cle.sort_values(by='Date', ascending=True)
print(sp500_cle.iloc[[0, -1]])
```

Data Analisis and modelling

```python
#sp500 weekly return
sp500_cle['Return'] = round((sp500_cle['Adj Close'].pct_change())*100, 3)
```

```python
mean_sp500 = round(sp500_cle['Return'].mean(),3)
std_sp500 = round(sp500_cle['Return'].std(ddof = 0),3)
median_sp500 = round(sp500_cle['Return'].median(),3)

qua_low_sp500 = round(sp500_cle['Return'].quantile(0.025),3)
qua_hight_sp500 = round(sp500_cle['Return'].quantile(0.975),3)

sharpe_ratio = round(mean_sp500/std_sp500,3)
```

```python
annual_yield_direct = (
    sp500_cle.groupby('Year')['Adj Close']
    .apply(lambda x: ((x.iloc[-1] / x.iloc[0]) - 1) * 100)
    .round(3)
)

print(annual_yield_direct)
```

Data Storitelling

```python
print(f'The Mean is {mean_sp500}% \n'
      f'with {std_sp500}% standard deviation \n'
      f'the median is {median_sp500}% \n'
      f'while the interval of 95% is {qua_low_sp500}% and {qua_hight_sp500}%, \n'
      f'In the end the sharpe ratio is {sharpe_ratio}')
```
